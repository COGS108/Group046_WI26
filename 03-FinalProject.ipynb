## Summary of Our Findings
Across all datasets, the clearest and most consistent pattern is the seasonality in visitation instead of environmental decline. In summer months, vistor counts signiticantly peak (July-August ~500K-600K median) and drop significantly during winter months such as January and December (~100K-150K) which indicates that visitation is primarily driven by predictable seasonal factors rather than long-term environmental changes. While ozone air quality measurements show a slight improvement over time (−0.0006 ppm/year), the temperature of Yosemite remains relatively stable with a weak upward trend. This suggests that large-scale environmental shifts are not strongly reflected in the visitation patterns observed.

## Connection to Background Research
Our results do not necessarily contradict the prior literature, but they do suggest that our project measured another version of the problem. Prior studies such as Keiser, Lade, and Rudik and Minehart et al. suggest that pollution and air quality can reduce park visitation, especially when examined at broader scales or with more targeted pollution measures such as PM2.5. Our analysis supports the idea that environmental quality matters in principle, but the data available for this project may not have captured the most behaviorally relevant type of air pollution. In Yosemite, wildfire smoke is likely one of the most important environmental deterrents to outdoor recreation, yet wildfire smoke is better represented by PM2.5 than by ozone. As a result, our findings are better interpreted as showing that the specific environmental indicators available here were not strong predictors of visitation, rather than proving that environmental conditions do not matter.

## Visitor Behavior Patterns
Yosemite's visitor behavior is structured around seasonal cycles, as summer months have the highest median visitation but also relatively lower coefficients of variation (CV ≈ 0.16-0.19), reflecting consistent demand on a year to year basis. For example, August has both the highest mean (~562,476 visitors) and one of the lowest relative variabilities, reinforcing the summer season as a stable peak season. 

In contrast, winter months such as December (CV ≈ 0.32) and February (CV ≈ 0.25) display significantly higher variability relative to their means. This potentially suggests that winter visitation is more sensitive to external factors such as weather conditions, holidays, and accessibility restrictions. When a park has such a strong recurring tourism cycle, gradual environmental variation may be overshadowed by the influence of school schedules, vacation planning, and other factors that are heavily influenced by seasonal recreation patterns. 

We then examined the extent to which AQI affects visitation through in-depth analysis. In our final Analysis, we observed correlation coefficients between visitors and avg_monthly_aqi, avg_monthly_tmax, and avg_monthly_prcp of 0.78, 0.92, and -0.53 respectively. However, upon performing a regression, we observed the avg_monthly_aqi to have a p-value of 0.73>0.05, making it not staistically significant. In contrast, both avg_monthly_tmax and avg_monthly_prcp had p-values<0.05, making them acceptable predictors of visitation rates. 

## Limitations
Several limitations constrained the conclusions we drew. Although the project was framed around daily AQI and daily trailhead counts, the visitation data available was only monthly. As such, daily data for the other variables needed to be reinterpreted in the context of monthly data, which reduced the ability of the independent and control variables to accurately represent the conditions present. Secondly, environmental measurements are imperfect proxies for actual conditions experienced across Yosemite's diverse terrain; EPA and NOAA measurements come from fixed monitoring locations and may not accurately reflect conditions at high-elevation or remote trailheads.

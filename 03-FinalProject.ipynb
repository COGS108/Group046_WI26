## Abstract

## Research Question

To what extent does variation in daily Air Quality Index (AQI), as measured by the U.S. Environmental Protection Agency (EPA), explain variation in daily visitor trailhead sensor counts in Yosemite National Park between 1990 and 2025, controlling for daily maximum temperature and total precipitation?

## Background and Prior Work

Outdoor recreation in national parks is influenced by environmental conditions that affect both visitor safety and overall experience. In the western United States, air quality has become an increasingly important factor due to wildfire smoke and regional pollution. The U.S. Environmental Protection Agency’s Air Quality Index (AQI) is a standardized measure used to communicate the health risks associated with air pollution and to inform the public about outdoor activity safety.

Because many national park activities involve extended outdoor physical activity, changes in air quality may affect visitors’ decisions about whether and when to visit. Understanding how air quality relates to park visitation is therefore relevant for park management, public health communication, and planning in highly visited parks such as Yosemite National Park.

Previous research has established a clear connection between environmental conditions and tourism behavior. Buckley (2011) reviews existing literature on tourism and the environment and finds that weather, climate variability, and environmental quality consistently influence visitation patterns in natural areas. <a href="#ref1">1</a> The review also notes that visitor responses differ depending on the type of destination and visitor motivations, suggesting that environmental effects may not be uniform across parks. However, Buckley’s work is conceptual and broad in scope, leaving open questions about how specific environmental indicators, such as AQI, influence visitation at individual parks.

Empirical studies provide further evidence that air pollution affects park visitation. Keiser, Lade, and Rudik (2018) examine monthly visitation data from 33 U.S. national parks and find that higher ozone pollution levels are associated with reduced visitation, even after accounting for weather, seasonality, and park-specific differences. <a href="#ref2">2</a> While this study demonstrates that air pollution can influence park use at a national scale, it relies on monthly averages and does not isolate park-specific daily behavioral responses. In addition, although Yosemite is included in their broader sample, the study does not focus specifically on Yosemite’s unique visitation patterns or wildfire-related air quality variability.

More recent work has used alternative data sources to study visitation patterns at finer time scales. Minehart et al. (2025) analyze weekly visitation to parks and protected areas in the Pacific Northwest using anonymized mobile device data, examining the effects of temperature, precipitation, AQI, and particulate matter (PM2.5). <a href="#ref3">3</a> The authors find that higher levels of PM2.5 are generally associated with lower visitation, while the relationship between AQI and visitation varies across different types of parks. However, their analysis aggregates across multiple parks and regions, making it difficult to determine how daily air quality fluctuations affect a single, high-profile destination over a long time horizon.

Together, these studies suggest that environmental quality and weather conditions play an important role in shaping park visitation, while also highlighting important gaps in the literature. Much of the prior work relies on monthly or weekly data and emphasizes cross-park comparisons rather than park-specific, daily analysis. Furthermore, limited attention has been given to how short-term variation in AQI influences visitation in an iconic and highly visited park such as Yosemite, particularly over multiple decades that include periods of increasing wildfire activity.

The present project addresses these gaps by examining whether daily variation in AQI predicts daily visitation to Yosemite National Park from 1979 to 2023, while controlling for daily maximum temperature and total precipitation. By focusing on a single park and using daily data, this study aims to capture short-term behavioral responses to air quality conditions that may be obscured in broader, aggregated analyses. This park-specific approach provides clearer insight into how air pollution influences visitation decisions in one of the most prominent national parks in the United States.

```python
import pandas as pd
import os

# load dataset
url = "https://raw.githubusercontent.com/COGS108/Group046_WI26/master/data/00-raw/conreport_1990_2025.csv"
aqi_raw = pd.read_csv(url)

# clean column names
aqi_raw.columns = aqi_raw.columns.str.strip().str.lower()

# filter Mariposa County
aqi = aqi_raw[aqi_raw['county'] == 'Mariposa County, CA'].copy()

# rename important columns
aqi = aqi.rename(columns={
    "year": "Year",
    "county": "County",
    "county code": "County Code",
    "pm2.5 98th percentile 24-hr": "PM2.5 98th Pct",
    "pm2.5 weighted mean 24-hr": "PM2.5 Mean",
    "ozone 4th max 8-hr": "Ozone 8hr Max",
    "ozone 2nd max 1-hr": "Ozone 1hr Max",
    "co 2nd max 8-hr": "CO 8hr Max",
    "no2 mean 1-hr": "NO2 Mean",
    "pm10 mean 24-hr": "PM10 Mean"
})

# convert pollutants to numeric
pollutant_cols = [col for col in aqi.columns if col not in ['Year', 'County Code', 'County']]
for col in pollutant_cols:
    aqi[col] = pd.to_numeric(aqi[col], errors='coerce')

# clean  column
aqi["Year"] = pd.to_numeric(aqi["Year"], errors="coerce")
aqi = aqi.dropna(subset=["Year"])
aqi["Year"] = aqi["Year"].astype(int)

# remove duplicate years
aqi = aqi.drop_duplicates(subset=["Year"]).sort_values("Year").reset_index(drop=True)

print("Final AQI dataset shape:", aqi.shape)
print("Year range:", aqi["Year"].min(), "-", aqi["Year"].max())

# save cleaned dataset
os.makedirs("data/02-processed", exist_ok=True)
aqi.to_csv("data/02-processed/mariposa_aqi.csv", index=False)

print("Clean AQI data saved.")
```

### Data Wrangling Summary

The visitation data was downloaded from the National Park Service statistics portal and cleaned before analysis.

Cleaning steps included:

- Removing non-data header rows from the raw file
- Dropping unnecessary columns
- Parsing the date column into a usable datetime format
- Renaming columns for clarity
- Converting visitor counts to numeric values

The cleaned dataset was then saved for further analysis and merged with the environmental datasets.

---

### Shortcomings

- **Monthly resolution:** The data is only available at a monthly level rather than daily, which limits the ability to observe short-term behavioral responses.

- **Limited sample size:** With 564 observations, statistical power may be lower compared to daily datasets.

```python
import pandas as pd
import io
import requests
import os

#load data
url = "https://raw.githubusercontent.com/COGS108/Group046_WI26/72efcfcfe6ea229d895a0e93db5f02fc74ef297c/data/00-raw/Recreation%20Visitors%20By%20Month%20(1979%20-%20Last%20Calendar%20Year).csv"
s = requests.get(url).content

#data cleaning (getting rid of non-data rows and columns)
#remove "LabelName" and "Yosemite NP" lines and "Textbox5" column
df = pd.read_csv(io.StringIO(s.decode('utf-8')), skiprows=2)
df = df.drop(columns=['Textbox5'])

#wrangle columns
month_cols = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
for col in month_cols:
    df[col] = df[col].astype(str).str.replace(',', '').replace('nan', '0')
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)

# switch to reading lengthwise instead of widthwise
df_tidy = df.melt(id_vars=['Year'],
                  value_vars=month_cols,
                  var_name='Month',
                  value_name='Visitors')

month_map = {'JAN':1, 'FEB':2, 'MAR':3, 'APR':4, 'MAY':5, 'JUN':6,
             'JUL':7, 'AUG':8, 'SEP':9, 'OCT':10, 'NOV':11, 'DEC':12}
df_tidy['Month_Num'] = df_tidy['Month'].map(month_map)
df_tidy['Date'] = pd.to_datetime(df_tidy[['Year']].assign(Month=df_tidy['Month_Num'], Day=1))
#exclude 2026
df_tidy = df_tidy[df_tidy['Year'] != 2026].copy()

# 1. Size of dataset
print(f'Shape: {df_tidy.shape}')
print(f'Date range: {df_tidy["Date"].min().date()} to {df_tidy["Date"].max().date()}')

# 2. Tidiness check
print(f'\nDuplicate month/year combos: {df_tidy[["Year", "Month"]].duplicated().sum()}')

# 3. Missingness
print('\nMissing values per column:')
print(df_tidy.isnull().sum())

# 4. Missing calendar months
# Check for gaps between 1979 and the most recent year
expected_months = pd.date_range(start='1979-01-01', end=df_tidy['Date'].max(), freq='MS')
missing_months = expected_months.difference(df_tidy['Date'])
print(f'\nCalendar months with no visitor data: {len(missing_months)}')
if len(missing_months) > 0:
    print(f'Missing periods: {missing_months}')

# 5. Outliers
print('\nVisitor count summary statistics:')
print(df_tidy['Visitors'].describe())
# outlier: exceptionally low month (<50k visitors)
low_traffic = df_tidy[df_tidy['Visitors'] < 50000]
print(f'\nMonths with < 50,000 Visitors: {len(low_traffic)}')
print(low_traffic[['Date', 'Visitors']])

# 6. Seasonality breakdown
avg_by_month = df_tidy.groupby('Month')['Visitors'].mean().sort_values(ascending=False)
print('\nAverage Visitors by Month (Historical):')
print(avg_by_month)

# 7. Save cleaned file
os.makedirs('data/02-processed', exist_ok=True)
df_tidy.to_csv('data/02-processed/yosemite_visitors_cleaned.csv', index=False)
print('\nSaved to data/02-processed/yosemite_visitors_cleaned.csv')

#sort values
df_tidy = df_tidy.sort_values('Date').reset_index(drop=True)

print(df_tidy.head())
```

```python
import pandas as pd
import os
from IPython.display import display

# load weather data (Yosemite)
path = "https://raw.githubusercontent.com/COGS108/Group046_WI26/vy/data/00-raw/yosemite_weather_1990_2024.csv"
weather = pd.read_csv(path)

# fix DATE column to datetime
weather['DATE'] = pd.to_datetime(weather['DATE'], errors='coerce')

# important columns are numeric
weather['TMAX'] = pd.to_numeric(weather['TMAX'], errors='coerce')
weather['PRCP'] = pd.to_numeric(weather['PRCP'], errors='coerce')

# dataset before cleaning
print("before cleaning:")
display(pd.DataFrame({
    "rows": [weather.shape[0]],
    "columns": [weather.shape[1]]
}))

print("date range:")
display(pd.DataFrame({
    "start": [weather["DATE"].min()],
    "end": [weather["DATE"].max()]
}))

print("duplicate dates:")
display(pd.DataFrame({
    "count": [weather["DATE"].duplicated().sum()]
}))

print("missing values (important columns):")
display(weather[['TMAX', 'PRCP']].isnull().sum().to_frame(name="missing"))

print("stats before cleaning:")
display(weather[['TMAX', 'PRCP']].describe())

# fix unrealistic temperature values
weather.loc[(weather['TMAX'] > 115) | (weather['TMAX'] < -30), 'TMAX'] = pd.NA

# drop rows where date is missing
weather = weather.dropna(subset=['DATE'])

# remove duplicate dates and sort everything
weather = (
    weather
    .drop_duplicates(subset=['DATE'])
    .sort_values('DATE')
    .reset_index(drop=True)
)

# after cleaning
print("after cleaning:")
display(pd.DataFrame({
    "rows": [weather.shape[0]],
    "columns": [weather.shape[1]]
}))

print("missing values after cleaning:")
display(weather[['TMAX', 'PRCP']].isnull().sum().to_frame(name="missing"))

print("stats after cleaning:")
display(weather[['TMAX', 'PRCP']].describe())

# save cleaned dataset
os.makedirs('data/02-processed', exist_ok=True)
weather.to_csv('data/02-processed/yosemite_weather_clean.csv', index=False)

print("saved cleaned weather data")
```

```python
import pandas as pd
import os

# Load raw precipitation data
path = "https://raw.githubusercontent.com/COGS108/Group046_WI26/Ria/data/00-raw/raw_precipitation_data.csv"
df = pd.read_csv(path)

# 1. Clean column names
df.columns = df.columns.str.strip()

# 2. Filter Yosemite data
precip = df[df['STATION_NAME'].str.contains('YOSEMITE', na=False)].copy()

# 3. Rename + parse date
precip['Date'] = pd.to_datetime(precip['DATE'], errors='coerce')
precip = precip.rename(columns={'PRCP': 'Precipitation'})

# 4. Basic dataset info
print("Shape:", precip.shape)
print("Date range:", precip['Date'].min(), "to", precip['Date'].max())

# 5. Check duplicates
print("\nDuplicate dates:", precip['Date'].duplicated().sum())

# 6. Missing values
print("\nMissing values per column:")
print(precip.isnull().sum())

# 7. Missing calendar days
full_range = pd.date_range(start=precip['Date'].min(), end=precip['Date'].max())
missing_days = full_range.difference(precip['Date'])

print("\nCalendar days with no precipitation reading:", len(missing_days))
print("Missing days are dropped (not filled), since no record ≠ zero precipitation.")

# 8. Summary statistics
print("\nPrecipitation summary statistics:")
print(precip['Precipitation'].describe())

# 9. Outliers (extreme precipitation days)
outliers = precip[precip['Precipitation'] > 100]
print("\nExtreme precipitation days:", len(outliers))

# 10. Monthly totals (for trends)
precip['Month'] = precip['Date'].dt.to_period('M')
monthly_totals = precip.groupby('Month')['Precipitation'].sum()

print("\nTop 10 wettest months:")
print(monthly_totals.sort_values(ascending=False).head(10))

# 11. Save cleaned dataset
os.makedirs('data/02-processed', exist_ok=True)
precip.to_csv('data/02-processed/yosemite_precipitation.csv', index=False)

print("\nSaved to data/02-processed/yosemite_precipitation.csv")
```

### 1. Air Quality Trends (Ozone)

We analyze air quality in Mariposa County using ozone levels as our primary metric. Ozone is consistently recorded across all years and is a major contributor to overall AQI in the Yosemite region, making it a reliable measure for long-term analysis.

To examine changes over time, we plot annual ozone levels with a linear trend line to capture both year-to-year variability and any overall trend.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# load processed data
aqi = pd.read_csv('data/02-processed/mariposa_aqi.csv')

# ensure year is integer
aqi['Year'] = pd.to_numeric(aqi['Year'], errors='coerce')
aqi = aqi.dropna(subset=['Year'])
aqi['Year'] = aqi['Year'].astype(int)

# use Ozone 8hr Max as primary air quality measure
ozone = aqi[['Year', 'Ozone 8hr Max']].dropna().reset_index(drop=True)

# fit linear trend line
z = np.polyfit(ozone['Year'], ozone['Ozone 8hr Max'], 1)
p = np.poly1d(z)

# create figure
fig, ax = plt.subplots(figsize=(12, 5))

# scatter plot
ax.scatter(
    ozone['Year'],
    ozone['Ozone 8hr Max'],
    s=60,
    zorder=3,
    label='Annual Ozone 4th Max 8-hr'
)

# trend line
x_range = np.linspace(ozone['Year'].min(), ozone['Year'].max(), 200)
ax.plot(
    x_range,
    p(x_range),
    linestyle='--',
    linewidth=2,
    label=f'Trend (slope={z[0]:.4f} ppm/year)'
)

# annotate worst year
worst = ozone.loc[ozone['Ozone 8hr Max'].idxmax()]
ax.annotate(
    f"{int(worst['Year'])}\n({worst['Ozone 8hr Max']:.3f} ppm)",
    xy=(worst['Year'], worst['Ozone 8hr Max']),
    xytext=(worst['Year'] - 3, worst['Ozone 8hr Max'] + 0.005),
    arrowprops=dict(arrowstyle='->')
)

# labels and formatting
ax.set_xlabel('Year')
ax.set_ylabel('Ozone 4th Max 8-hr (ppm)')
ax.set_title('Annual Ozone Levels - Mariposa County, CA (1990–2025)',
             fontweight='bold')
ax.xaxis.set_major_locator(ticker.MultipleLocator(5))
ax.legend()

plt.tight_layout()
plt.savefig('data/02-processed/ozone_trend.png', dpi=150, bbox_inches='tight')
plt.show()

# print summary
best = ozone.loc[ozone['Ozone 8hr Max'].idxmin()]

print(f"Overall trend: {z[0]:+.4f} ppm per year")
print(f"Worst year:  {int(worst['Year'])} ({worst['Ozone 8hr Max']:.3f} ppm)")
print(f"Best year:   {int(best['Year'])} ({best['Ozone 8hr Max']:.3f} ppm)")
```

The plot shows that ozone levels vary noticeably from year to year, with several spike years indicating periods of poorer air quality. These spikes likely correspond to environmental factors such as wildfire activity or extreme heat.

The fitted trend line suggests a slight overall decrease in ozone levels over time, though the change is small. This indicates that while long-term air quality may not be consistently worsening, short-term fluctuations remain significant.

The annotated worst year highlights a period with particularly high ozone levels, while the lowest point represents relatively cleaner air conditions. These differences emphasize the variability of air quality across years.

Overall, air quality in Mariposa County shows variability rather than a strong consistent trend. These fluctuations are important for our analysis, as periods of poor air quality may impact visitor experience and behavior in Yosemite.

### 2. Visitor Patterns

We use Yosemite National Park visitation data from the National Park Service (NPS) as our primary measure of visitor behavior. This dataset contains monthly visitation counts from 1979 to 2025, totaling 564 observations.

This dataset serves as our key dependent variable, representing how park attendance changes over time. Because the data is recorded monthly, it allows us to analyze long-term trends and seasonal patterns in visitation.

One limitation of this dataset is that visitation is only recorded at the monthly level rather than daily. This reduces the resolution of the data and means short-term fluctuations cannot be directly observed.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

#import data
df = pd.read_csv(('data/02-processed/yosemite_visitors_cleaned.csv'), index_col='Date', parse_dates=True)

df['Month'] = df.index.month_name()

#plot data
plt.figure(figsize=(12, 6))
sns.boxplot(x='Month', y='Visitors', data=df,
            order=['January', 'February', 'March', 'April', 'May', 'June',
                   'July', 'August', 'September', 'October', 'November', 'December'])
plt.title('Distribution of Visitors by Month (1979-2025)')
plt.xticks(rotation=45)
plt.show()

#---------------
# Load and prepare data
df = pd.read_csv('data/02-processed/yosemite_visitors_cleaned.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').set_index('Date')

# Global Variance and Distribution Stats
global_stats = df['Visitors'].agg(['mean', 'median', 'var', 'std', 'skew'])

# Monthly Volatility, Grouping by month to see seasonal variance
# Coeff of var CV = std / mean
monthly_stats = df.groupby(df.index.month_name())['Visitors'].agg(['mean', 'var', 'std'])
monthly_stats['CV'] = monthly_stats['std'] / monthly_stats['mean']

# 3.variation by decade
df['Decade'] = (df.index.year // 10) * 10
decade_stats = df.groupby('Decade')['Visitors'].agg(['mean', 'var', 'std'])
decade_stats['CV'] = decade_stats['std'] / decade_stats['mean']

# 4.YoY % Change
annual_totals = df['Visitors'].resample('A').sum()
yoy_growth = annual_totals.pct_change()
growth_variance = yoy_growth.var()

# Display Results
print("Global Summary Statistics:")
print(global_stats)
print("\nMonthly Volatility (Sorted by CV):")
print(monthly_stats.sort_values('CV', ascending=False))
```

The distribution of visitors by month shows a clear seasonal pattern. Visitation peaks during the summer months (June through August) and is lowest during the winter months, reflecting typical travel behavior and weather conditions in Yosemite.

The boxplot also indicates that peak months have greater variability, suggesting that visitation during these periods is more sensitive to external factors such as weather or environmental conditions.

Overall, Yosemite visitation follows a strong seasonal pattern with higher and more variable visitor counts during peak months. These patterns provide a baseline for our analysis, as environmental factors may influence when and how often people visit the park.

### 3. Temperature Trends

We analyze daily maximum temperature (TMAX) data from NOAA to understand long-term climate patterns in the Yosemite region. Temperature is an important factor because it can influence both air quality and visitor comfort.

This dataset provides high-frequency daily observations, allowing for a more detailed view of variability compared to monthly datasets.


```python
import pandas as pd
import os
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Configuration
sns.set_theme(style="whitegrid", font_scale=1.2)
plt.rcParams['figure.figsize'] = (10, 6)

# 2. Data grabbing
path = 'https://raw.githubusercontent.com/COGS108/Group046_WI26/vy/data/00-raw/yosemite_weather_1990_2024.csv'
weather = pd.read_csv(path)
print(" Raw weather data loaded.\n")

# 3. Structural cleaning and creating column names
weather.columns = weather.columns.str.strip().str.upper()

# Convert date and numeric columns
weather['DATE'] = pd.to_datetime(weather['DATE'], errors='coerce')
for col in ['TMAX', 'TMIN', 'PRCP']:
    if col in weather.columns:
        weather[col] = pd.to_numeric(weather[col], errors='coerce')

# 4. Data diagnostics
print("--- DATASET ARCHITECTURE ---")
print(f"Total Observations: {weather.shape[0]}")
print(f"Temporal Scope:     {weather['DATE'].min().date()} to {weather['DATE'].max().date()}")
print(f"Duplicate Entries:  {weather['DATE'].duplicated().sum()}")

print("\n--- MISSINGNESS AUDIT ---")
existing_cols = [col for col in ['TMAX','TMIN','PRCP'] if col in weather.columns]
print(weather[existing_cols].isnull().sum())

# 5. Outlier
if 'TMAX' in weather.columns:
    logic_mask = (weather['TMAX'] > 115) | (weather['TMAX'] < -30)
    extreme_temps = weather[logic_mask]
    print(f"\nImprobable TMAX days found: {len(extreme_temps)}")

if 'TMIN' in weather.columns:
    logic_mask_min = (weather['TMIN'] > 100) | (weather['TMIN'] < -40)
    extreme_tmin = weather[logic_mask_min]
    print(f"Improbable TMIN days found: {len(extreme_tmin)}")

# 6. Visual Exploratory Data Analysis
if 'TMAX' in weather.columns:
    plt.figure()
    sns.histplot(weather['TMAX'], bins=30, kde=True, color="teal")
    plt.title("Distribution of Yosemite Daily Max Temperatures (1990-2024)")
    plt.xlabel("Temperature (TMAX)")
    plt.ylabel("Frequency")
    plt.axvline(115, color='red', linestyle='--', label='Upper Bound')
    plt.axvline(-30, color='red', linestyle='--', label='Lower Bound')
    plt.legend()
    plt.show()

if 'PRCP' in weather.columns:
    plt.figure()
    sns.histplot(weather['PRCP'].dropna(), bins=30, kde=False, color="skyblue")
    plt.title("Distribution of Daily Precipitation (PRCP)")
    plt.xlabel("Precipitation (inches)")
    plt.ylabel("Frequency")
    plt.show()

# 7. Persistence
os.makedirs('data/02-processed', exist_ok=True)
weather.to_csv('data/02-processed/yosemite_weather_clean.csv', index=False)
print("\n Cleaned weather dataset saved to: data/02-processed/yosemite_weather_clean.csv")
```

The distribution of daily maximum temperatures appears approximately normal, with most values falling within a realistic range. Extreme values are relatively rare and were addressed during data cleaning.

Over time, temperature shows noticeable variability across years, along with a slight upward trend. This suggests gradual warming in the region over the study period.

Overall, temperature patterns indicate increasing variability and a modest long-term rise. These trends are relevant to our analysis, as higher temperatures may influence both air quality conditions and visitor behavior.

### 4. Precipitation Trends

We analyze total precipitation (PRCP) data from NOAA to examine rainfall patterns in Yosemite. Precipitation is important because it affects environmental conditions such as wildfire risk, trail accessibility, and overall visitor experience.

Like temperature, this dataset provides daily observations, allowing us to capture detailed variability over time.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# load raw data
path = "https://raw.githubusercontent.com/COGS108/Group046_WI26/Ria/data/00-raw/raw_precipitation_data.csv"
df = pd.read_csv(path)

# filter Yosemite station
precip = df[df['STATION_NAME'].str.contains('YOSEMITE', na=False)].copy()

# correct date conversion
precip['Date'] = pd.to_datetime(precip['DATE'], format='%Y%m%d')

# rename precipitation column
precip.rename(columns={'PRCP': 'Precipitation'}, inplace=True)

# extract year after fixing date
precip['Year'] = precip['Date'].dt.year

# aggregate to annual totals
annual_total = (
    precip
    .groupby('Year')['Precipitation']
    .sum()
    .reset_index()
)

# fit linear trend line
z = np.polyfit(annual_total['Year'],
               annual_total['Precipitation'], 1)
p = np.poly1d(z)

# create figure
fig, ax = plt.subplots(figsize=(12, 5))

# scatter plot
ax.scatter(
    annual_total['Year'],
    annual_total['Precipitation'],
    s=60,
    zorder=3,
    label='Annual Total Precipitation'
)

# trend line
x_range = np.linspace(
    annual_total['Year'].min(),
    annual_total['Year'].max(),
    200
)

ax.plot(
    x_range,
    p(x_range),
    linestyle='--',
    linewidth=2,
    label=f'Trend (slope={z[0]:.2f} inches/year)'
)

# annotate wettest year
wettest = annual_total.loc[
    annual_total['Precipitation'].idxmax()
]

ax.annotate(
    f"{int(wettest['Year'])}\n({wettest['Precipitation']:.1f} in)",
    xy=(wettest['Year'], wettest['Precipitation']),
    xytext=(wettest['Year'] - 3,
            wettest['Precipitation'] + 5),
    arrowprops=dict(arrowstyle='->')
)

# labels & formatting
ax.set_xlabel('Year')
ax.set_ylabel('Total Annual Precipitation (inches)')
ax.set_title('Annual Total Precipitation - Yosemite (1990–2025)',
             fontweight='bold')

ax.xaxis.set_major_locator(ticker.MultipleLocator(5))

plt.tight_layout()
plt.show()

# print summary
driest = annual_total.loc[
    annual_total['Precipitation'].idxmin()
]

print(f"Overall trend: {z[0]:+.3f} inches per year")
print(f"Wettest year: {int(wettest['Year'])} "
      f"({wettest['Precipitation']:.1f} inches)")
print(f"Driest year: {int(driest['Year'])} "
      f"({driest['Precipitation']:.1f} inches)")
```

Precipitation shows high variability across time, with some periods experiencing significantly higher or lower rainfall. These fluctuations reflect natural climate variability, including wet and dry years.

There is no strong consistent long-term trend, but extreme precipitation events stand out clearly, indicating periods of unusually high or low rainfall.

Overall, precipitation in Yosemite is highly variable rather than steadily increasing or decreasing. This variability is important for understanding environmental conditions that may influence both air quality and visitor patterns.
